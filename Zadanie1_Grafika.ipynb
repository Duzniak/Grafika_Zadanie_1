```python
import urllib.request
import cv2
import matplotlib.pyplot as plt

url = "https://commons.wikimedia.org/wiki/Special:FilePath/Fresh_(96795219).jpeg?width=800"
req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})

with urllib.request.urlopen(req) as response:
  with open("Fresh.jpeg", "wb") as file:
    file.write(response.read())

img = cv2.imread("Fresh.jpeg")

plt.figure(figsize=(6,4))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Oryginał")
plt.axis("off")
plt.show()

new_width = int(img.shape[1] * 0.5)
new_height = int(img.shape[0] * 0.5)

img_resized = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_AREA)
img_gray = cv2.cvtColor(img_resized, cv2.COLOR_BGR2GRAY)

img_rotated = cv2.rotate(img_gray, cv2.ROTATE_90_CLOCKWISE)

plt.figure(figsize=(4,6))
plt.imshow(img_rotated, cmap="gray")
plt.title("Obrócone")
plt.axis("off")
plt.show()

print("Wymiary macierzy (wysokość, szerokość):", img_rotated.shape)
print("Macierz obrazu:\n", img_rotated)
```
